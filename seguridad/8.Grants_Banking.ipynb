#Grants: administración de permisos de Unity Catalog

Accesos:
- DataEngineers: uso del catálogo, acceso de lectura/escritura y creación de tablas en todo el catálogo.
- DataAnalysts: acceso de solo lectura al esquema golden.
- user_projecto_banking1@outlook.com y user_projecto_banking2@outlook.com: mismos permisos que DataEngineers.

```python
# Catálogo

dbutils.widgets.text("catalog_name","banking_dev","Nombre del catálogo")
catalog_name = dbutils.widgets.get("catalog_name").strip()

if not catalog_name:
    raise ValueError("El parámetro catalog_name no puede estar vacío")

print(f"Catálogo: {catalog_name}")
```

**Permisos de ingeniería**  
Se conceden a DataEngineers y a los dos usuarios individuales.  
- En el catálogo: USE CATALOG, CREATE SCHEMA, SELECT y MODIFY.
- En cada esquema (raw, bronze, silver, golden): USE SCHEMA y CREATE TABLE.
Los permisos SELECT y MODIFY a nivel de catálogo se heredan en los objetos descendientes según el modelo de Unity Catalog.

```python
%sql
GRANT USE CATALOG, CREATE SCHEMA, SELECT, MODIFY
ON CATALOG `${catalog_name}`
TO `DataEngineers`;

GRANT USE CATALOG, CREATE SCHEMA, SELECT, MODIFY
ON CATALOG `${catalog_name}`
TO `user_projecto_banking1@outlook.com`;

GRANT USE CATALOG, CREATE SCHEMA, SELECT, MODIFY
ON CATALOG `${catalog_name}`
TO `user_projecto_banking2@outlook.com`;
```

```python
%sql
GRANT USE SCHEMA, CREATE TABLE
ON SCHEMA `${catalog_name}`.`raw`
TO `DataEngineers`;

GRANT USE SCHEMA, CREATE TABLE
ON SCHEMA `${catalog_name}`.`bronze`
TO `DataEngineers`;

GRANT USE SCHEMA, CREATE TABLE
ON SCHEMA `${catalog_name}`.`silver`
TO `DataEngineers`;

GRANT USE SCHEMA, CREATE TABLE
ON SCHEMA `${catalog_name}`.`golden`
TO `DataEngineers`;
```

```python
%sql
GRANT USE SCHEMA, CREATE TABLE
ON SCHEMA `${catalog_name}`.`raw`
TO `user_projecto_banking1@outlook.com`;

GRANT USE SCHEMA, CREATE TABLE
ON SCHEMA `${catalog_name}`.`bronze`
TO `user_projecto_banking1@outlook.com`;

GRANT USE SCHEMA, CREATE TABLE
ON SCHEMA `${catalog_name}`.`silver`
TO `user_projecto_banking1@outlook.com`;

GRANT USE SCHEMA, CREATE TABLE
ON SCHEMA `${catalog_name}`.`golden`
TO `user_projecto_banking1@outlook.com`;
```

```python
%sql
GRANT USE SCHEMA, CREATE TABLE
ON SCHEMA `${catalog_name}`.`raw`
TO `user_projecto_banking2@outlook.com`;

GRANT USE SCHEMA, CREATE TABLE
ON SCHEMA `${catalog_name}`.`bronze`
TO `user_projecto_banking2@outlook.com`;

GRANT USE SCHEMA, CREATE TABLE
ON SCHEMA `${catalog_name}`.`silver`
TO `user_projecto_banking2@outlook.com`;

GRANT USE SCHEMA, CREATE TABLE
ON SCHEMA `${catalog_name}`.`golden`
TO `user_projecto_banking2@outlook.com`;
```

**Permisos para analistas**  <br>  
DataAnalysts podrá utilizar el catálogo y consultar las tablas del esquema golden. No se le conceden permisos de escritura ni acceso a los esquemas raw, bronze y silver.

```python
%sql
GRANT USE CATALOG
ON CATALOG `${catalog_name}`
TO `DataAnalysts`;

GRANT USE SCHEMA, SELECT
ON SCHEMA `${catalog_name}`.`golden`
TO `DataAnalysts`;
```
