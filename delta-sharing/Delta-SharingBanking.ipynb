##Delta Sharing – PROD a Power BI  <br><br>
Proyecto: Banking Analytics Lakehouse  
Catálogo: banking_prod | Esquema: golden  
Share: banking_analytics_share  
Recipient: powerbi_banking_prod

Este notebook configura Delta Sharing Databricks-to-Open mediante Bearer Token para que Power BI consuma exclusivamente las tablas agg de Golden.



```sql
SELECT current_catalog() AS catalogo_actual, current_schema() AS esquema_actual;

SHOW TABLES IN banking_prod.golden;
```

```sql
-- CREAR EL SHARE

CREATE SHARE IF NOT EXISTS banking_analytics_share;
```

```sql
-- AGREGAR TABLAS AL SHARE

ALTER SHARE banking_analytics_share
ADD TABLE banking_prod.golden.agg_fraud_daily;

ALTER SHARE banking_analytics_share
ADD TABLE banking_prod.golden.agg_sbs_creditos;

ALTER SHARE banking_analytics_share
ADD TABLE banking_prod.golden.agg_sbs_depositos;

ALTER SHARE banking_analytics_share
ADD TABLE banking_prod.golden.agg_sbs_morosidad;
```

```sql
SHOW ALL IN SHARE banking_analytics_share;
```

```sql
-- CREAR RECIPIENT PARA POWER BI

CREATE RECIPIENT IF NOT EXISTS powerbi_banking_prod
COMMENT 'Consumidor Power BI de las tablas agregación Golden en PROD';
```

```sql
-- CONCEDER ACCESOS AL SHARE

GRANT SELECT ON SHARE banking_analytics_share
TO RECIPIENT powerbi_banking_prod;
```

```sql
SHOW GRANTS ON SHARE banking_analytics_share;
```

```sql
SHOW SHARES;
SHOW RECIPIENTS;
SHOW ALL IN SHARE banking_analytics_share;
```

```sql
-- OBTENER ENLACE DE ACTIVACIÓN

DESCRIBE RECIPIENT powerbi_banking_prod;
```
